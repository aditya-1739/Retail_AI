```python
import json

with open("/content/drive/MyDrive/SmartRetailAI/datasets/intents.json") as file:
    data = json.load(file)
```

```python
patterns = []
tags = []

responses = {}

for intent in data["intents"]:
    responses[intent["tag"]] = intent["responses"]

    for pattern in intent["patterns"]:
        patterns.append(pattern)
        tags.append(intent["tag"])
```

```python
print(patterns)
print(tags)
```

```python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(patterns)

y = tags
```

```python
model = LogisticRegression()

model.fit(X, y)
```

```python
joblib.dump(
    model,
    "/content/drive/MyDrive/SmartRetailAI/models/chatbot_model.pkl"
)

joblib.dump(
    vectorizer,
    "/content/drive/MyDrive/SmartRetailAI/models/chatbot_vectorizer.pkl"
)
```

```python
def chatbot(message):

    message_vector = vectorizer.transform([message])

    prediction = model.predict(message_vector)[0]

    return responses[prediction][0]
```

```python
print(chatbot("Hi"))
```

```python
print(chatbot("Store timing"))
```

```python
print(chatbot("Do you accept UPI?"))
```

```python
while True:

    message = input("You: ")

    if message.lower() == "quit":
        print("Bot: Goodbye!")
        break

    print("Bot:", chatbot(message))
```
